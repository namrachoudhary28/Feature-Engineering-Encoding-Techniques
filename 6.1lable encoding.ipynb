# lable encoding

```python
import pandas as pd
```

```python
df = pd.DataFrame({"name":["sql", "python", "excel", "pandas", "ml", "dl"]})
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder

# sklearn library h pandas numpy ki trha / jise import krna ho
```

```python
# ab ise ek variable me store kr do

le = LabelEncoder()    
```

```python
# fit transform ki help se jo text vala (name) vala coulmn 
# hota h use no. me transfer krta h 

le.fit_transform(df["name"])
```

```python
# ab ise bhi ek variable me store kr do 

df["en_name"] = le.fit_transform(df["name"])
```
